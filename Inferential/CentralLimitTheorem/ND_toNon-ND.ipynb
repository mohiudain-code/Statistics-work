```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pop_data=[np.random.randint(1,300) for i in range(1,5000)]
len(pop_data)
```

```python
df=pd.DataFrame({'pop_data':pop_data})
df.head()
```

```python
plt.figure(figsize=(7,4))
sns.kdeplot(x='pop_data',data=df)
plt.show()
```

```python
F_sample=[]
for no_sample in range(50):
    sample=[]
    for i in range(500):
        sample.append(np.random.choice(df['pop_data']))
    F_sample.append(np.mean(sample))
```

```python
df=pd.DataFrame({'F_Mean':F_sample})
df.head()
```

```python
plt.figure(figsize=(7,4))
sns.kdeplot(x='F_Mean',data=df)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dict={
    'Name':['Jillani','Qaisar','Anas','Ali','Ahmad','Zohaib','Jawad'],
    'Age':[20,20,21,22,19,20,23],
    'Distance':[20,32,10,26,20,18,56],
    'Time':[30,50,20,35,30,27,10]
}
df=pd.DataFrame(dict)
df.head()
```

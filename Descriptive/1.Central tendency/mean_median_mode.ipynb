```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("train_update.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df['MoSold'].mean()
```

```python
avg=np.sum(df['MoSold'])/len(df['MoSold'])
print(avg)
```

```python
median=df['MoSold'].median()
```

```python
mode=df['MoSold'].mode()[0]
```

```python
sns.histplot(x='MoSold',data=df)
plt.plot([avg for i in range(0,350)],[i for i in range(0,350)],c='green',label='mean')
plt.plot([median for i in range(0,350)],[i for i in range(0,350)],c='gray',label='median')
plt.plot([mode for i in range(0,350)],[i for i in range(0,350)],c='red',label='mode')
plt.legend()
plt.show()
```

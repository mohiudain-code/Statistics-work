```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=np.random.normal(0,50,50)
```

```python
df=pd.DataFrame({'Rnd':df})
df.head()
```

```python
df.skew()['Rnd']
```

```python
sns.histplot(x='Rnd',data=df)
plt.show()
```

```python
df['Rnd'].mean(),df['Rnd'].mode()[0],df['Rnd'].median()
```

Mean Absolute Deviation

```python
Sec_A=np.array([65,76,70,67,72])
Sec_B=np.array([89,56,80,68,74])
count=np.array([1,2,3,4,5])
```

```python
mA=Sec_A.mean()
mB=Sec_B.mean()
mB
```

```python
plt.scatter(Sec_A,count,label='Sec_A')
plt.scatter(Sec_B,count,label='Sec_B')
plt.plot([70,70,70,70,70],count,c='red',label='mean')
plt.plot([73,73,73,73,73],count,c='green',label='mean')
plt.legend()
plt.show()
```

```python
MAD=np.sum(np.abs(Sec_A-mA))/len(Sec_A)
MAD
```

```python
MAD=np.sum(np.abs(Sec_B-mB))/len(Sec_B)
MAD
```

```python
np.std(Sec_A)
np.std(Sec_B)
```
